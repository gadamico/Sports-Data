```python
import pandas as pd
import sqlite3
```

Here's a fairly natural way to calculate the best men's and women's players for a given year. Simply give it to the player with the most victories in major tournaments. If there's a tie, then break the tie by giving it to the player, among those tied, with the most major *finals*. If there's still a tie, then look at major *semifinals*, etc. If two or more players are still tied, then we can start looking at performance in the Masters 1000 series. Let's work this out for men and women in the Open Era. (See the notebook in the womens_tennis folder for the latter.)

General note in what follows: I'll list what round of the majors was relevant to determining each year's "winner", but it doesn't follow from that indication that the winner was the *only* player to achieve that mark. For example, Stan Smith made it to three major quarterfinals in 1971, but there may have been others to achieve that. Smith is the best player (by our metric) of 1971 not simply because he made three major quarters but because he:

- won as many majors as anyone in 1971; and
- made it to as many finals in majors as anyone; and
- made it to as many semis in majors as anyone; and
- made it to more quarters in majors than anyone else.

## Building a Data Table

Let's put this data in a pair of dictionaries, one for the men and one for the women.

```python
moy_dict = {}

moy_dict['Rod Laver'] = [1968, 1969]
moy_dict['Ken Rosewall'] = [1970]
moy_dict['Stan Smith'] = [1971]
moy_dict['Ilie Nastase'] = [1972]
moy_dict['John Newcombe'] = [1973]
moy_dict['Jimmy Connors'] = [1974, 1982]
moy_dict['Bjorn Borg'] = [1975, 1976, 1978, 1979, 1980]
moy_dict['Guillermo Vilas'] = [1977]
moy_dict['John McEnroe'] = [1981, 1984]
moy_dict['Mats Wilander'] = [1983, 1988]
moy_dict['Ivan Lendl'] = [1985, 1986, 1987]
moy_dict['Boris Becker'] = [1989, 1991]
moy_dict['Stefan Edberg'] = [1990]
moy_dict['Jim Courier'] = [1992]
moy_dict['Pete Sampras'] = [1993, 1994, 1995, 1996, 1997, 1998, 2000]
moy_dict['Andre Agassi'] = [1999, 2001]
moy_dict['Lleyton Hewitt'] = [2002]
moy_dict['Juan Carlos Ferrero'] = [2003]
moy_dict['Roger Federer'] = [2004, 2005, 2006, 2007, 2009]
moy_dict['Rafael Nadal'] = [2008, 2010, 2013, 2017, 2019, 2022]
moy_dict['Novak Djokovic'] = [2011, 2012, 2014, 2015, 2016, 2018, 2021, 2023]
moy_dict['Dominic Thiem'] = [2020]
moy_dict['Jannik Sinner'] = [2024, 2025]
```

```python
df = pd.DataFrame([(key, len(val)) for key, val in moy_dict.items()], columns=['Player', 'MOYs'])
df2 = df.sort_values(by='MOYs', ascending=False)
```

```python
df2 = df2.reset_index(drop=True)
df2
```

```python
con = sqlite3.connect('../data/tennis.db')
```

```python
df2.to_sql('moy', con=con)
```
